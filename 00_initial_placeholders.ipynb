```python
import numpy as np
print("Numpy Imported")
```

```python
lin_space = np.linspace([0,0,0], [10,20,30], 10 ,axis=1)
#lin_space -> make a np array (start, stop, num=50, endpoint=True, retstep=False, dtype=None, axis=0)
#endpoint=True -> including the last origin(index:50) from the (num=50).
#retstep=False -> if you want to return only array, retstep=True -> return the steps also(lin_space,steps = np.linspace(0,5,num=50,retstep=True))
#dtype=None -> Type of the output values, Default: float64
#axis -> control the call index/column function
#and the shape organization rather insert as    axis=0 : (10, 3) ,     axis=1 : (3, 10)
l1 = lin_space[0]
l2 = lin_space[1]
l3 = lin_space[2]

print(f"l1:\n{l1}")
print(f"l2:\n{l2}")
print(f"l3:\n{l3}")
print(f"\nlin space:\n{lin_space.shape}")
```

```python
ara_nge = np.arange(0, 10, 2,dtype=float)
#ara_nge -> make a np array ([start,] stop, [step,] dtype=None, *, like=None)
#start -> Default = 0
#step -> jump size between origins
#like -> return the new array as type as the input array(GPU,NumPy,CPU)

print(f"\narrange:\n{ara_nge}")
```

```python
zeros_arr = np.zeros((2,),dtype=int,like=ara_nge)
ones_arr = np.ones((2,),dtype=int,like=ara_nge)
#zeros_arr -> make a np array od values=0 (shape, dtype, order='C'/'F', *, like=None)
#shape >= 1D writen with () as tuple if more then 1 Dimension.
print(f"\nzeros_arr:\n{zeros_arr}")
print(f"\nones_arr:\n{ones_arr}")
```

```python
full_arr = np.full((2,2),7)
# fill the entire new array/matrix with specific value. (shape, fill_value, dtype=None, order='C', *, like=None)
print(f"full_arr:\n{full_arr}")
```

```python
eye_arr = np.eye(N=4,M=5,k=1)
# fill the new matrix's slant with value = 1 and all other fields with value = 0 (N, M=None, k=0, dtype=float, order='C', *, like=None)
#N -> number of rows if N=4 matrix.shape = (4,4) #if you want to change the column count you add optional M=count (Default:None)
#k -> declare where the start point of the matrix's slant.(Optional,Default=None)
print(f"eye_arr:\n{eye_arr}")
```

```python
random_arr = np.random.random(size=(2,2))
# always return [float]/matrix[float] unless size=None in this case the function return float.
print(f"random_arr:\n{random_arr}")
```

```python
empty_arr = np.empty(shape=(3,5))
#this function reusing the "garbage" values of the existing memory anf fill the new array/matrix with them.
# the values in this out-put is an accident because in the previous cell we declare arr of zeros, we can't
# start from an assumption that the values will be only 0. if you put shape=(20,20) probably the values won't be 0.
#       (shape, dtype=float, order='C', *, like=None)
print(f"empty_arr:\n{empty_arr}")
```
