```python
import numpy as np
import os

BASE_DIR = os.getcwd()
print("Numpy&os Imported")
```

```python
a = [1, 2, 3, 4, 5, 6]

array_name = "my_array.npy"

np.save(array_name, a)
PATH_DIR_A = os.path.join(BASE_DIR,array_name)
print(PATH_DIR_A)
# saving numpy array as binary file (file, arr, allow_pickle=True, fix_imports=True)
# you can send list in the arr field and the function will automatically converty to 'numpy.ndarray'
# allow_pickle=True -> Default this is how you can save complex object not only array/matrix
# but when you load it to another file you need to declare in the load function allow_pickle=True,
# allow_pickle=False -> will prohibit saving complex objects as (dict,PersonalObject())


if os.path.exists(PATH_DIR_A):
    print(f"a Saved Successfully\n\ra:\n{a}")
```

```python
x = [1, 2, 3, 4, 5, 6]
f = {
    "key": "value" }
c = {
    "key2": "value2" }
saved_dict = {
    "first":x,
    "second":f,
    "third":c
}
saved_name = 'multi_array.npz'
SAVED_PATH = os.path.join(BASE_DIR,saved_name)
print(SAVED_PATH)
np.savez(saved_name,**saved_dict ,allow_pickle=True)
# save a couple of arrays in one file (file, *args, **kwargs)
# file -> path
# *args -> unnamed arrays the function generate automatically names(arr_0,arr_1...)
# **kwargs -> named arrays (better way to handle)
if os.path.exists(SAVED_PATH):
    print(f"saved_dict objects Saved Successfully\n\r{saved_dict}")
```

```python
PATH_MY_ARRAY = os.path.join(BASE_DIR,"my_array.npy")
arr = np.load(PATH_MY_ARRAY,allow_pickle=True)
#load the array.npz/npy files (file, mmap_mode=None, allow_pickle=False, fix_imports=True, encoding='ASCII')
# mmap_mode=None -> Default load all to ram memory, ="r":only reading from the disk not loading to the ram,
#                   ="+r": reading and writing from the disk. ="+w": only writing to disk, Use-full for reduce speed
#                   with big files.
#after loading npz we need to write data.close() for better performance and keeping the file secure.
print(arr)

```

```python
PATH_MY_ARRAY = os.path.join(BASE_DIR,"multi_array.npz")
arr = np.load(PATH_MY_ARRAY,allow_pickle=True)
#load the array.npz/npy files (file, mmap_mode=None, allow_pickle=False, fix_imports=True, encoding='ASCII')
# mmap_mode=None -> Default load all to ram memory, ="r":only reading from the disk not loading to the ram,
#                   ="+r": reading and writing from the disk. ="+w": only writing to disk, Use-full for reduce speed
#                   with big files.
#after loading npz we need to write data.close() for better performance and keeping the file secure.
print(dict(arr))
```

```python
TXT_NEW_PATH = os.path.join(BASE_DIR,"txt_for_load.txt")
TXT_ORIGINAl_PATH = os.path.join(os.getcwd(),"trying.txt")

with open(TXT_ORIGINAl_PATH, "r", encoding="utf-8") as f:
    original_text = f.read()
    f.seek(0)
    f.close()

lines = original_text.split("\n")

safe_lines = [line if line != "" else "no_line" for line in lines]

try:
    np.savetxt(TXT_NEW_PATH ,np.array(safe_lines, dtype=object).reshape(-1, 1),fmt="%s",delimiter="§")
    print(f"{TXT_NEW_PATH} Saved Successfully")

except Exception as e:
    print("Saving ERROR! :  ",str(e))

try:
    data = np.loadtxt(TXT_NEW_PATH,dtype=str,delimiter="§",encoding="utf-8")
    print(f"{TXT_NEW_PATH} Loaded Successfully")

    restored_data = [str(data)] if data.ndim == 0 else list(data)

    restored_lines = ["" if line == "no_line" else line for line in restored_data]

    restored_text = "\n".join(restored_lines)

    print("restored_text == original_text: -> ", restored_text == original_text)

except Exception as e:
    print("Loading ERROR! :  ",str(e))
```
