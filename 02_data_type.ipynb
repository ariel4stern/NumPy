```python
import numpy as np
print("Imported")
```

```python
arr = np.arange(10,dtype=np.int64)
# Complete numbers
print(arr[:3])
```

```python
arr = np.arange(10,dtype=np.float32)
# not accuracy like np.float64 but good for machine learning, 32 version is faster.
print(arr[:3])
```

```python
c = np.array([1+2j, 3+4j], dtype=np.complex128)
# complex number j^2 = -1 ,j represent the point in the 2D plain (1,2)(3,4)
# j is the spinning of the vector in the plain every j multiplication the vector spin 90 degrees negate the clock.
print(c, c.dtype)
```

```python
d = np.array([True, False, True], dtype=np.bool)
print(d, d.dtype)
```

```python
e = np.array([1, "hello", 3.5], dtype=object)
print(e, e.dtype)
```

```python
f = np.array(["hi", "hello"], dtype=np.str_)
print(f, f.dtype)
```

```python
g = np.array(["hi", "hello"], dtype=np.bytes_)
print(g, g.dtype)
```
