```python
a =  np.array([1,2,3,4])
b = np.array([
    [10,2,3,4]
    ,[5,6,7,8]
])
print("a : \n",a)
print("len(a) = ",len(a))
print("b : \n",b)
print("len(b) = ",len(b))
#Length of array samples/indexes/rows
```

```python
b = np.array([
    [10,2,3,4]
    ,[5,6,7,8]
])
b.ndim
#Number of array dimensions
```

```python
b = np.array([
    [10,2,3,4]
    ,[5,6,7,8]
])
b.dtype
#Data type of array elements
```

```python
b = np.array([
    [10,2,3,4]
    ,[5,6,7,8]
])
b.dtype.name
#Name of data type
```

```python
b = np.array([
    [10,2,3,4]
    ,[5,6,7,8]
])
b = b.astype(bool)
print("b.astype(bool):",b.dtype.name)

b = b.astype(float)
print("b.astype(float):",b.dtype.name)
# Convert an array to a different type
```
