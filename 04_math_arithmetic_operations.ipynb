```python
log = np.array([70,20,19,18,3,4,5,6,82,100],dtype=float)
l = np.log(log)
# b:  [1 2 3 4 5 6]
# the opposite of np.exp(),
# for x = 1
# np.exp calculate e^1 = (result : ~2.71828183).
# and np.log calculate e^(result : 0) = 1
# np.log(b) = [e^(?) =1 ,e^(?) =2 ,e^(?) =3 ,e^(?) =4 ,e^(?) =5 ,e^(?) =6]
print("log: ",log)

print(f"np.log(log) = l = \n",l)

reversed_l = np.exp(l)

print("reversed_l: ",reversed_l)
print("all(reversed_l==log) = ",True if all(reversed_l==log) else False)
print("np.allclose(reversed_l, log) = ",np.allclose(reversed_l, log))
# float type dangerous for (==) operation , it is checking to the last byts
# np.allclose()/np.isclose() check the float type if it is close enough to be called True
```

```python
angles = np.array([0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi])
sin_angles = np.sin(angles)
# angle = 0 : sin(0) = y(height) = ( ? (x,width)  , 0)
# π = 180 degrees in python (radians)
# RANGE OF THE RESULTS: -1 <= np.sin(x) <= 1
print("angles: ",np.round(angles,4))
print("np.sin(angles) = ", np.round(sin_angles,4))
```

```python
cos_angles = np.cos(angles)
# angle = 0 : sin(0) = y(height) = ( ? (x,width)  , 0)
# π = 180 degrees in python (radians)
# RANGE OF THE RESULTS: -1 <= np.sin(x) <= 1
print("angles: ",np.round(angles,4))
print("np.cos(angles) = ", np.round(cos_angles,4))

x = np.round(sin_angles,4)
y = np.round(cos_angles,4)
print(f"x,y:\n\t","\n\t".join([f"{float(x[i]), float(y[i])}" for i in range(len(x))]))
```
